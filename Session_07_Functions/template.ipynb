### Section 3 - Passing List to Functions

```python
def looping_list(items):
    for item in items:
        print (item)

testing_list = [1, 2, 3, 4, 5]
looping_list(testing_list)
```

### Section 4 - Example Program

```python
my_list = []

def printing_menu():
    print("Menu")
    print("=================")
    print("1. Add List")
    print("2. Show List")
    print("3. Exit")

def addList():
    num = int(input("Input number: "))
    my_list.append(num)

def showList():
    print("My List:")
    for item in my_list:
        print (item)

def main():
    while (True):
        printing_menu()
        action = int(input(">> "))
        if action == 1:
            addList()
        elif action == 2:
            showList()
        elif action == 3:
            break; 
        else:
            print("\nAction invalid!\n")

main()
```
